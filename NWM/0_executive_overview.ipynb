<center><h1>NWM API Overview</h1></center>

The National Water Model (NWM) is a hydrologic modeling framework that simulates observed and forecast streamflow over the entire continental United States (CONUS), southern Alaska (Cook Inlet, Copper River Basin, and Prince William Sound regions), Hawaii, Puerto Rico and the US Virgin Islands. Additionally, it produces total water level guidance for the coastlines of those same regions except Alaska. The NWM simulates the water cycle with mathematical representations of the different processes and how they fit together. The NWM produces hydrologic guidance at a very fine spatial and temporal scale. It complements official NWS river forecasts and produces guidance at millions of other locations that do not have a traditional river forecast.

This API provides access to the National Water Model (NWM) version 3.1 streamflow and total water level forecasts.  It supports six configurations: short_range, medium_range, medium_range_blend, long_range, analysis_assim, and analysis_assim_extend.

Currently, the API serves the most recent 3 days of forecast data, with the exception of the long_range configuration, which is limited to the past 24 hours.

The following data can be obtained from the NWM API:

* National Water Model reach metadata
* Short range streamflow forecasts (CONUS, AK, HI, PRVI)
* Medium range streamflow forecasts (CONUS, AK)
* Medium range blend streamflow forecasts (CONUS, AK)
* Long range streamflow forecasts (CONUS)
* Analysis Assim (CONUS, AK, HI, PRVI)
* Analysis Assim Extend (CONUS, AK)

* Short range total water level forecasts
* Short range psurge total water level forecasts
* Medium range total water level forecasts
* Medium range psurge total water level forecasts
* Medium range blend total water level forecasts
* Medium range blend psurge total water level forecasts
* Analysis Assim total water level forecasts
* Analysis Assim Extend total water level forecasts

The NWM API provides several service endpoints to facilitate efficient querying of the data of interest. See the table below for descriptions of each API endpoint:


| Streamflow Endpoints (/v1/streamflow) | Description |
| --- | --- |
| /configurations | Retrieves a list of available National Water Model (NWM) streamflow forecast configurations.<br>**Behavior:** <br>* *Without parameters:* Returns all available NWM configurations that have active streamflow forecasts.<br>* *With a specified configuration:* Returns a list of all NWM reaches associated with that specific configuration.<br>|
| /forecast | This endpoint retrieves National Water Model (NWM) streamflow forecasts.  Supports single or multiple comma-delimited NWM reaches, reference time filtering (specific time or range), and a latest=true flag to return only the most recent forecast.  Data is available in JSON or SHEF format; the PEDTSEP parameter is strictly required when requesting SHEF output. |
| /locations | Retrieves a list of National Water Model (NWM) reaches with available forecast data.<br>**Behavior:** <br>* *Default:* Returns a complete list of all NWM reaches that currently have available streamflow forecast data. <br>* *With a specified reach:* Returns the complete metadata for that specific NWM reach.<br>|

| Total Water Level Endpoints (/v1/total_water_level) | Description |
| --- | --- |
| /configurations | Retrieves a list of available National Water Model (NWM) total water level forecast configurations.<br>**Behavior:** <br>* *Without parameters:* Returns all available NWM configurations that have active total water level forecasts.<br>* *With a specified configuration:* Returns a list of all NWS gauges associated with that specific configuration.<br>|
| /forecast | This endpoint retrieves National Water Model (NWM) total water level forecasts.  Supports single or multiple comma-delimited NWS Location IDs, reference time filtering (specific time or range), and a latest=true flag to return only the most recent forecast.  Data is available in JSON or SHEF format; the PEDTSEP parameter is strictly required when requesting SHEF output. |
| /locations | Retrieves a list of NWS Location IDs with available total water level forecast data.<br>**Behavior:** <br>* *Default:* Returns a complete list of all NWS gauges that currently have available total water level forecast data. <br>* *With a specified reach:* Returns the metadata for that specific NWS gauge.<br>|
