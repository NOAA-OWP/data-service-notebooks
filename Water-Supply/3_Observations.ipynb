### Returns all of the recorded observations for a given location

Use this endpoint to inspect and filter observations:

* **Filter Reaches by Configuration:** Pass a specific location id to return observation values associated with it.
* **NOTE:** Because all historical records are retained, omitting time-range query parameters may result in large responses and degraded performance.


```python
# start by importing the requests library
import requests
# import the pprint library to make the output more readable
from pprint import pprint
# define the api-endpoint
API_ENDPOINT = "https://api.staging.water.noaa.gov/water-supply/v1"
```

```python
# set nws location id
nws_location_id = "CDLC1"
# create a observations request
uri = API_ENDPOINT + "/observations/" + nws_location_id + "/"
print(f'Retrieving observations of {nws_location_id} from {uri} ...')
# get the response
response = requests.request("GET", uri)
# print the response
pprint(response.json())
```

### Returns all of the recorded observations for a given location over a set time frame

We can also narrow down observations to a set time frame. Below we do the same call as below but only for a the month of april 2026. This is achieved by adding an `observation_type` parameter to the uri as well as a `valid_datetime_before` and `valid_datetime_after`.

```python
# set nws location id
nws_location_id = "CDLC1"
valid_datetime_before = "2026-04-30T12:00:00Z"
valid_datetime_after = "2026-04-01T12:00:00Z"
# create a observations request
uri = API_ENDPOINT + "/observations/" + nws_location_id + "/" + "?observation_type=&valid_datetime_before=" + valid_datetime_before + "&valid_datetime_after=" + valid_datetime_after
print(f'Retrieving observations of {nws_location_id} from {uri} ...')
# get the response
response = requests.request("GET", uri)
# print the response
pprint(response.json())
```
