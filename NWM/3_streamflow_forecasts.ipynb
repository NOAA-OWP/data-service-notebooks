## NWM Streamflow Forecasts API Parameters

When requesting data from the **Streamflow Forecasts** endpoints, you can pass parameters to filter model runs, specify timeframes, or set output options.

### Endpoint Parameters

| Parameter | Type | Required? | Description | Example / Valid Values |
| :--- | :--- | :---: | :--- | :--- |
| `select_by` | `string` | **Yes** | The location id type to query for NWM streamflow forecasts. | `nwm_feature_id`, `nws_location_id` |
| `selector` | `string` | **Yes** | The location id(s) to query for NWM streamflow forecasts. | `101`, `101,700694`, `HARP1`, `HARP1,BLUO2` |
| `configuration` | `string` | No | The configurations you'd like NWM streamflow forecasts for.  Can query for multiple configurations.  If no configuration is specified, all configurations are returned | `short_range`, `medium_range`, `medium_range_mean`, `medium_range_blend`, `long_range`, `long_range_mean`, `analysis_assim`, `analysis_assim_extend` |
| `reference_datetime` | `string` (ISO 8601) | No | Filters for a specific reference_datetime. | `2026-07-15T00:00:00Z` |
| `max_reference_datetime` | `string` (ISO 8601) | No | Includes forecast data where the reference time is on or prior to the `max_reference_datetime`. | `2027-08-30T23:00:00Z` |
| `min_reference_datetime` | `string` (ISO 8601) | No | Includes forecast data where the reference time is on or after the `after_reference_datetime`. | `2027-06-30T23:00:00Z` |
| `latest` | `boolean` | No | If set to true, only returns the latest forecasts. | `true`, `false` |
| `output_format` | `string` | No | Specify the output format.  Defaults to `json`.  **Note:** If set to shef, it will download the SHEF file. | `json`, `shef` |
| `pedtsep` | `string` | **No (Unless output_format is shef)** | If output_format is set to shef, the user must specify a PEDTSEP.  See page 7 in the <a href="https://www.weather.gov/media/mdl/SHEF_CodeManual_5July2012.pdf"> SHEF manual</a> for more details.| `QRIRGZ`, `HPIRGZ` |

### Retrieve the latest streamflow forecast for all configurations for 1 NWM reach

````python
# start by importing the requests library
import requests
# import the pprint library to make the output more readable
from pprint import pprint
# define the api-endpoint
API_ENDPOINT = "https://api.water.noaa.gov/nwm"
````

````python
select_by = 'nwm_feature_id'
nwm_feature_id = '101'
latest = 'true'

# create a streamflow configurations request
uri = f'{API_ENDPOINT}/v1/streamflow/forecast/{select_by}/{nwm_feature_id}?latest={latest}'

print(f'Retrieving short_range streamflow forecast for nwm_feature_id {nwm_feature_id} from {uri} ...')
# get the response
response = requests.request("GET", uri)
# print the response
pprint(response.json())
````
