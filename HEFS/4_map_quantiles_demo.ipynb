### Map Quantiles Overview

The Map Quantiles endpoint serves a set of quantiles based on the maximum value for each member of the ensemble forecast for a given forecast length. The `forecast_length` indicates the length of the map quantile forecast in days. The maximum value for each member of the ensemble forecast is determined for a given forecast length. For example, if 5 is the forecast length, then the maximum values for each ensemble member are selected from the first 5 days of the forecast only. The api will return a quantile set for each predetermined forecast length (5, 10, 15, 20, 30, 60, 90, or 120 days). Quantile sets are calculated up to the maximum length that the ensemble forecast can support. For example, a 25-day ensemble forecast will only include map quantiles for lengths of 5, 10, 15, and 20 days. The content of `quantile_values` corresponds to the exceedance quantile values in the order given by the `exceedance_quantiles` in the metadata. In the example below, an exceedance number of `0.1` with the value of `566.73` cfs, means that only 10% of ensemble members in this forecast have a maximum that exceeds `566.73` cfs.

  ### Retrieving Data from the API

Like with the hydrograph quantiles example, We first import Python's requests library, which allows us to make HTTP requests. We then define the API endpoint we'll be interacting with, setting API_ENDPOINT to https://api.water.noaa.gov/hefs/. Here's how we do it:

````python
# start by importing the requests library
import requests
# import the pprint library to make the output more readable
from pprint import pprint
# import pandas for easier quantile calculations and formatting
import pandas as pd
# import scipy for quantile calculations
from scipy.stats.mstats import mquantiles

# define the api-endpoint
API_ENDPOINT = "https://api.water.noaa.gov/hefs"
# define location id
LOCATION_ID = "SSAC1"
# define parameter id
PARAMETER_ID = "QINE"
````

Now, we will grab the map quantiles for the specified location id and parameter id. To do this we will use the map quantiles endpoint as follows:

`/v1/map-quantiles/?location_id={LOCATION_ID}&parameter_id={PARAMETER_ID}`

Note: Threshold data is not available for many locations so will often be displayed as none.

````python
# create a series request with location_id, parameter_id, start_date_date, and limit filters
uri = API_ENDPOINT + f"/v1/map-quantiles/?location_id={LOCATION_ID}&parameter_id={PARAMETER_ID}"
# get the response
response = requests.request("GET", uri)
# print the response
pprint(response.json())
````
