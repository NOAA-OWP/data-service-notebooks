## NWM Total Water Level Forecasts API Parameters

When requesting data from the **Total Water Level Forecasts** endpoints, you can pass parameters to filter model runs, specify timeframes, or set output options.

### Endpoint Parameters

| Parameter | Type | Required? | Description | Example / Valid Values |
| :--- | :--- | :---: | :--- | :--- |
| `select_by` | `string` | **Yes** | The location id type to query for NWM streamflow forecasts. | `nws_location_id` |
| `selector` | `string` | **Yes** | The location id(s) to query for NWM streamflow forecasts. | `THOM1`, `THOM1,ALBN6` |
| `configuration` | `string` | No | The configurations you'd like NWM streamflow forecasts for.  Can query for multiple configurations.  If no configuration is specified, all configurations are returned | `short_range`, `medium_range`,  `medium_range_blend`, `analysis_assim`, `analysis_assim_extend`, `short_range_psurge`, `medium_range_psurge`, `medium_range_blend_psurge` |
| `reference_datetime` | `string` (ISO 8601) | No | Filters for a specific reference_datetime. | `2026-07-15T00:00:00Z` |
| `before_reference_datetime` | `string` (ISO 8601) | No | Filters for all forecasts before the `before_reference_datetime`. | `2027-08-30T23:00:00Z` |
| `after_reference_datetime` | `string` (ISO 8601) | No | Filters for all forecasts after the `before_reference_datetime`. | `2027-06-30T23:00:00Z` |
| `latest` | `boolean` | No | If set to true, only returns the latest forecasts. | `true`, `false` |
| `output_format` | `string` | No | Specify the output format.  Defaults to `json`.  **Note:** If set to shef, it will download the SHEF file. | `json`, `shef` |
| `pedtsep` | `string` | **No (Unless output_format is shef)** | If output_format is set to shef, the user must specify a PEDTSEP. | `QRIRGZ`, `HPIRGZ` |

### Retrieve the latest total water level forecast for all configurations for 1 NWS gauge

````python
# start by importing the requests library
import requests
# import the pprint library to make the output more readable
from pprint import pprint
# define the api-endpoint
API_ENDPOINT = "https://api.staging.water.noaa.gov/nwm"
````

````python
select_by = 'nws_location_id'
nws_location_id = 'THOM1'
latest = 'true'

# create a total_water_level configurations request
uri = f'{API_ENDPOINT}/v1/total_water_level/forecast/{select_by}/{nws_location_id}?latest={latest}'

print(f'Retrieving short_range total_water_level forecast for nws_location_id {nws_location_id} from {uri} ...')
# get the response
response = requests.request("GET", uri)
# print the response
pprint(response.json())
````
