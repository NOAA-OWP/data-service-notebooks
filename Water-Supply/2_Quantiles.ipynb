### Returns all quantiles generated for a specific location for any arbitrary time range

A water supply forecast is an ensemble prediction of the total volume of streamflow at a gauge over an extended period of time, such several months. An "exceedance quantile" of a given percentage (e.g., 25%, 50%, 75%), then, is the volume of water that has that percentage change of being exceeded based on the ensemble water supply forecast. For example, if the water supply forecast period is from Apr 1 through July 31 at a given gauge location, and the 25% exceedance quantile is 1000 KAF, then there is a 25% chance that the total volume of streamflow at that gauge from Apr 1 to July 31 exceeds 1000 KAF.

Use this endpoint to inspect quantiles:

* **Filter Reaches by Configuration:** Pass a specific location id to return quantile values associated with it.
* **NOTE:** Because all historical records are retained, omitting time-range query parameters may result in large responses and degraded performance.


```python
# start by importing the requests library
import requests
# import the pprint library to make the output more readable
from pprint import pprint
# define the api-endpoint
API_ENDPOINT = "https://api.staging.water.noaa.gov/water-supply/v1"
```

```python
# set nws location id
nws_location_id = "CDLC1"
# create a observations request
uri = API_ENDPOINT + "/quantiles/" + nws_location_id + "/"
print(f'Retrieving quantiles of {nws_location_id} from {uri} ...')
# get the response
response = requests.request("GET", uri)
# print the response
pprint(response.json())
```

### Returns all quantiles generated for a specific location for a specific time range and forecast period uuid

We can also obtain a quantile with more specificty. We can request for a specific time range and forecast period. This involves adding the parameters to the uri `valid_datetime_before`, `valid_datetime_after`, and `forecast_period_uuid`.

```python
from typing import ForwardRef
# set nws location id
nws_location_id = "CDLC1"
valid_datetime_before = "2026-01-31"
valid_datetime_after = "2026-01-01"
forecast_period_uuid = "afebfd33-5c68-4246-a0e7-596d77fba8c4"
# create a observations request
uri = API_ENDPOINT + "/quantiles/" + nws_location_id + "/" + "?valid_datetime_before=" + valid_datetime_before + "&valid_datetime_after=" + valid_datetime_after + "&forecast_period_uuid=" + forecast_period_uuid
print(f'Retrieving quantiles of {nws_location_id} from {uri} ...')
# get the response
response = requests.request("GET", uri)
# print the response
pprint(response.json())
```
