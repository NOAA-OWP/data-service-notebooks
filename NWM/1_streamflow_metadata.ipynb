### Streamflow configurations endpoint

Use this endpoint to inspect and filter National Water Model (NWM) streamflow forecast configurations:

* **Retrieve All Configurations:** Fetch the complete list of available NWM streamflow forecast configurations.
* **Filter Reaches by Configuration:** Pass a specific configuration to return all NWM streamflow reaches associated with it.

````python
# start by importing the requests library
import requests
# import the pprint library to make the output more readable
from pprint import pprint
# define the api-endpoint
API_ENDPOINT = "https://api.water.noaa.gov/nwm"
````

````python
# create a streamflow configurations request
uri = API_ENDPOINT + "/v1/streamflow/configurations"

print(f'Retrieving streamflow configurations from {uri} ...')
# get the response
response = requests.request("GET", uri)
# print the response
pprint(response.json())
````

````python
# create a specific configuration request and retrieve the first 10 NWM reaches for that configuration
uri = API_ENDPOINT + "/v1/streamflow/configurations/short_range"
print(f'Retrieving the first 10 NWM reaches for a given configurations')

# get the response.  This response will retrieve all NWM reaches for the given configuration.  We will only print the first 10.
response = requests.request("GET", uri)
# print the response
pprint(response.json()['nwm_feature_ids'][:10])  # print the first 10 NWM feature IDs)
````
